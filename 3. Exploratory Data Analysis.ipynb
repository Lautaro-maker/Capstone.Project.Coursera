#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
miss_empl = df["Employment"].isnull().sum()
print(f"Missing values in Employment: {miss_empl}")

miss_jobsat = df["JobSat"].isnull().sum()
print(f"Missing values in JobSat: {miss_jobsat}")

miss_rw = df["RemoteWork"].isnull().sum()
print(f"Missing values in RemoteWork: {miss_rw}")
```

```python
#First, replace missing values in JobSat to the average of the column
df["JobSat"].dtype

import numpy as np
avg_jobsat = df["JobSat"].mean()
print(f"Most frequent value in JobSat: {avg_jobsat}")
df["JobSat"] = df["JobSat"].replace(np.nan, avg_jobsat)
#print(df["JobSat"].info())


#Secondly, replace in RemoteWork with the most frequent value (mode)

moda_rw= df["RemoteWork"].mode()[0]
print(f"Most frequent value in RemoteWork: {moda_rw}")

df["RemoteWork"] = df["RemoteWork"].fillna(moda_rw)
#print(df["RemoteWork"].info())
```

```python
miss_jobsat = df["JobSat"].isnull().sum()
print(f"Missing values in JobSat: {miss_jobsat}")

miss_rw = df["RemoteWork"].isnull().sum()
print(f"Missing values in RemoteWork: {miss_rw}")
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
#
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce').fillna(0)

YCP_bins = np.linspace(min(df["YearsCodePro"]), max(df["YearsCodePro"]), 5)

group_names = ["Junior", "Intermediate", "Senior", "Expert"]

df["YearsCodeProNew"] = pd.cut(df ["YearsCodePro"], bins = YCP_bins, labels = group_names, include_lowest= True)

df_grp = df.groupby("YearsCodeProNew", as_index = False)["JobSat"]. mean()                                                               ### Use .mean because a nosense result with the median

#Visualization

plt.figure(figsize=(10, 6))
sns.barplot(x="YearsCodeProNew", y="JobSat", data=df_grp, palette="viridis")

plt.title("Median Job Satisfaction by Experience Range")
plt.ylabel("JobSat Median")
plt.xlabel("Experience Level")
plt.show()
```

```python
#code with .median
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce').fillna(0)

YCP_bins = np.linspace(min(df["YearsCodePro"]), max(df["YearsCodePro"]), 5)

group_names = ["Junior", "Intermediate", "Senior", "Expert"]

df["YearsCodeProNew"] = pd.cut(df ["YearsCodePro"], bins = YCP_bins, labels = group_names, include_lowest= True)

df_grp = df.groupby("YearsCodeProNew", as_index = False)["JobSat"]. median()                                                               ### Code with median

#Visualization

plt.figure(figsize=(10, 6))
sns.barplot(x="YearsCodeProNew", y="JobSat", data=df_grp, palette="viridis")

plt.title("Median Job Satisfaction by Experience Range")
plt.ylabel("JobSat Median")
plt.xlabel("Experience Level")
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
df_lang = df[['Country', 'LanguageHaveWorkedWith']].dropna()
df_lang['Language'] = df_lang['LanguageHaveWorkedWith'].str.split(';')
df_exploded = df_lang.explode('Language')
df_exploded['Language'] = df_exploded['Language'].str.strip()

# Top 10 language in Top 5 countries
top_langs = df_exploded['Language'].value_counts().nlargest(10).index
top_countries = df['Country'].value_counts().nlargest(5).index
df_final_lang = df_exploded[df_exploded['Language'].isin(top_langs) & df_exploded['Country'].isin(top_countries)]

plt.figure(figsize=(14, 7))
sns.countplot(data=df_final_lang, x='Language', hue='Country', palette='magma')
plt.title("Top 10 most used languages ​​in the top 5 countries", pad=20)
plt.xticks(rotation=45)
plt.tight_layout(pad=3.0)
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ed_order = {
    'Primary/elementary school': 1,
    'Secondary school': 2,
    'Some college/university study': 3,
    'Associate degree': 4,
    'Bachelor’s degree': 5,
    'Master’s degree': 6,
    'Professional degree': 7,
    'Other doctoral degree': 8
}

emp_order = {
    'Employed, part-time': 1,
    'Independent contractor, freelancer, or self-employed': 2,
    'Employed, full-time': 3
}

df['EdLevel_Num'] = df['EdLevel'].map(ed_order)
df['Employment_Num'] = df['Employment'].map(emp_order)

df_counts_ed = df.groupby(['Employment', 'EdLevel']).size().reset_index(name='Count')
pivot_ed = df_counts_ed.pivot(index='Employment', columns='EdLevel', values='Count').fillna(0)

print(pivot_ed)

plt.figure(figsize=(10, 6))
sns.regplot(x='EdLevel_Num', y='Employment_Num', data=df, 
            x_jitter=0.2, y_jitter=0.2, 
            scatter_kws={'alpha':0.2}, 
            line_kws={'color':'red'})

plt.xticks(ticks=range(1, 9), labels=['Prim', 'Sec', 'Univ', 'Assoc', 'Bach', 'Mast', 'Prof', 'PhD'], rotation=45)
plt.yticks(ticks=[1, 2, 3], labels=['Part-time', 'Freelance', 'Full-time'])

plt.title("Correlación entre Nivel Educativo y Tipo de Empleo")
plt.xlabel("Nivel Educativo (Escala Numérica)")
plt.ylabel("Tipo de Empleo (Escala de Estabilidad)")
plt.grid(True, alpha=0.3)
plt.show()
```
