#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
print(df.dtypes)
print()

print(df.count())
print()

Miss_df = df.isnull().sum()
print(f"Missing Values:\n{Miss_df}")
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
country_counts = df['Country'].value_counts()
print(f"How many countries of each one: \n{country_counts}") 
print()
df_null_country = df['Country'].isnull().sum()
print(f"Missing Values: {df_null_country}") 
print()
df_nonull_country = df['Country'].count()
print(f"Valid Values: {df_nonull_country}") 
print()
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
mapping = {
    'United States of America': 'USA',
    'United Kingdom of Great Britain and Northern Ireland': 'UK',
    'Russian Federation': 'Russia'
}

df['Country'] = df['Country'].replace(mapping)

ed_mapping = {
    'Bachelor’s degree (B.A., B.S., B.Eng., etc.)': 'Bachelor',
    'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)': 'Master',
    'Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)': 'High School'
}

df['EdLevel'] = df['EdLevel'].replace(ed_mapping)
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
encoded_empl = pd.get_dummies(df["Employment"])
df = pd.concat([df, encoded_empl], axis=1)
print(df[['Employment'] + list(encoded_empl.columns)].head(2))
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
miss_df = df.isnull().sum().sort_values(ascending=False)
print(f"Cantidad de Valores Faltantes: \n{miss_df.head(10)}")
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
import numpy as np

miss_CCY= df["ConvertedCompYearly"].isnull().sum()
print(f"Cantidad de Valores Faltantes: {miss_CCY}")
avg_salary = df["ConvertedCompYearly"].mean()
print(f"El valor mas frecuente de ConvertedCompYearly es : {avg_salary}")
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].replace(np.nan, avg_salary)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
miss_RW = df["RemoteWork"].isnull().sum()
print(f"Cantidad de Valores Faltantes: {miss_RW}")
print()

moda_RW= df["RemoteWork"].mode()[0]
print(f"El valor mas frecuente de Employment es : {moda_RW}")
print()

df["RemoteWork"] = df["RemoteWork"].fillna(moda_RW)
print(df["RemoteWork"].isnull().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
df["ConvertedCompYearly_MinMax"]= (df["ConvertedCompYearly"] - df["ConvertedCompYearly"].min())/(df["ConvertedCompYearly"].max() - df["ConvertedCompYearly"].min())
print(df["ConvertedCompYearly_MinMax"].head(5))
print(df["ConvertedCompYearly_MinMax"].nunique())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
import matplotlib.pyplot as plt

df['ConvertedComp_Log'] = np.log1p(df['ConvertedCompYearly'])
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce').fillna(0)

def classify_experience(years):
    if years <= 2:
        return 'Junior'
    elif years <= 5:
        return 'Intermediate'
    elif years <= 10:
        return 'Senior'
    else:
        return 'Expert'

df['ExperienceLevel'] = df['YearsCodePro'].apply(classify_experience)

# Verificamos los resultados
print(df[['YearsCodePro', 'ExperienceLevel']].head(10))
print("\nConteo por nivel de experiencia:")
print(df['ExperienceLevel'].value_counts())
```
