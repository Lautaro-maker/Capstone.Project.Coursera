### Task 1: Exploring Data Distributions Using Bubble Plots


#### 1. Bubble Plot for Age vs. Frequency of Participation


- Visualize the relationship between respondents’ age and their participation frequency (`SOPartFreq`) using a bubble plot.


- Use the size of the bubbles to represent their job satisfaction (`JobSat`).


```python
order_age = [
    'Under 18 years old', 
    '18-24 years old', 
    '25-34 years old', 
    '35-44 years old', 
    '45-54 years old', 
    '55-64 years old', 
    '65 years or older', 
    'Prefer not to say'
]
order_freq = [
        "I have never participated in Q&A on Stack Overflow",
        "Less than once per month or monthly",
        "A few times per month or weekly",
        "A few times per week",
        "Daily or almost daily",
        "Multiple times per day"]


df['Age'] = pd.Categorical(df['Age'], categories=order_age, ordered=True)
df['SOPartFreq'] = pd.Categorical(df['SOPartFreq'], categories=order_freq, ordered=True)

df_plot = df.dropna(subset=['Age', 'SOPartFreq', 'JobSat'])
plt.figure(figsize=(10, 6))
plt.scatter(x = df_plot["Age"].cat.codes, y = df_plot['SOPartFreq'].cat.codes, s=df_plot['JobSat']*15, alpha=0.4, edgecolors="black")
plt.title("Relationship between respondents` age and their participation frequency")
plt.xlabel('Age')
plt.xticks(ticks=range(len(order_age)), labels=order_age, rotation=15)
plt.yticks(ticks=range(len(order_freq)),labels=order_freq)
plt.ylabel('Participation Frequency')
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Bubble Plots


#### 1. Bubble Plot for Compensation Across Developer Roles

- Visualize compensation (`ConvertedCompYearly`) across different developer roles (`DevType`).

- Use bubble size to represent job satisfaction.


```python
df_roles = df.dropna(subset=['DevType', 'ConvertedCompYearly', 'JobSat']).copy()

df_roles = df_roles[df_roles['ConvertedCompYearly'] < df_roles['ConvertedCompYearly'].quantile(0.95)]
df_roles['Role'] = df_roles['DevType'].str.split(';')
df_roles = df_roles.explode('Role')

df_comp_roles = df_roles.groupby('Role').agg({'ConvertedCompYearly': 'mean', 'JobSat': 'mean'}).reset_index()

plt.figure(figsize=(12, 8))
sns.scatterplot(data=df_comp_roles, x='ConvertedCompYearly', y='Role', size='JobSat', hue='Role', sizes=(100, 1000), alpha=0.7, legend=False)
plt.title('Compensation Across Developer Roles (Size = Job Satisfaction)')
plt.xlabel('Average Yearly Compensation')
plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Collaboration Tools by Age

- Visualize the relationship between the collaboration tools used (`NEWCollabToolsHaveWorkedWith`) and age groups.

- Use bubble size to represent the frequency of tool usage.


```python
orden_edad = ['Under 18 years old', '18-24 years old', '25-34 years old', '35-44 years old', '45-54 years old', '55-64 years old', '65 years or older', 'Prefer not to say']
df['Age'] = pd.Categorical(df['Age'], categories=orden_edad, ordered=True)

df_tools = df.dropna(subset=['NEWCollabToolsHaveWorkedWith', 'Age']).copy()
df_tools['Tool'] = df_tools['NEWCollabToolsHaveWorkedWith'].str.split(';')
df_tools = df_tools.explode('Tool')

df_counts_tools = df_tools.groupby(['Age', 'Tool']).size().reset_index(name='ToolUsage')

top_tools = df_tools['Tool'].value_counts().nlargest(10).index
df_counts_tools = df_counts_tools[df_counts_tools['Tool'].isin(top_tools)]

plt.figure(figsize=(12, 6))
sns.scatterplot(data=df_counts_tools, x='Age', y='Tool', size='ToolUsage', hue='Tool', sizes=(50, 2000), alpha=0.6, legend=False)

plt.title('Bubble Plot for Collaboration Tools by Age')
plt.xticks(rotation=45)
plt.xlabel('Age Groups')
plt.ylabel('Collaboration Tools')
plt.tight_layout()
plt.show()
```

### Task 4: Visualizing Technology Trends Using Bubble Plots


#### 1. Bubble Plot for Preferred Web Frameworks vs. Job Satisfaction

- Explore the relationship between preferred web frameworks (`WebframeWantToWorkWith`) and job satisfaction.

- Use bubble size to represent the number of respondents.



```python
df_web = df.dropna(subset=['WebframeWantToWorkWith', 'JobSat']).copy()
df_web['Framework'] = df_web['WebframeWantToWorkWith'].str.split(';')
df_web = df_web.explode('Framework')

df_counts_web = df_web.groupby(['Framework', 'JobSat']).size().reset_index(name='Respondents')

top_frameworks = df_web['Framework'].value_counts().nlargest(10).index
df_counts_web = df_counts_web[df_counts_web['Framework'].isin(top_frameworks)]

plt.figure(figsize=(12, 8))
sns.scatterplot(data=df_counts_web, x='JobSat', y='Framework', size='Respondents', hue='Framework', sizes=(50, 2000), alpha=0.6)
plt.title('Bubble Plot for Preferred Web Frameworks vs. Job Satisfaction')
plt.xlabel('Job Satisfaction Level')
plt.ylabel('Web Framework')
plt.xticks(rotation=15)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Frameworks')
plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Admired Technologies Across Countries

- Visualize the distribution of admired technologies (`LanguageAdmired`) across different countries (`Country`).

- Use bubble size to represent the frequency of admiration.



```python
df_admired = df.dropna(subset=['LanguageAdmired', 'Country']).copy()
df_admired['Language'] = df_admired['LanguageAdmired'].str.split(';')
df_admired = df_admired.explode('Language')

top_countries = df_admired['Country'].value_counts().nlargest(10).index
top_languages = df_admired['Language'].value_counts().nlargest(10).index

df_counts_admired = df_admired.groupby(['Language', 'Country']).size().reset_index(name='AdmirationFreq')

df_counts_admired = df_counts_admired[
    (df_counts_admired['Country'].isin(top_countries)) & 
    (df_counts_admired['Language'].isin(top_languages))
]

plt.figure(figsize=(12, 10))
sns.scatterplot(data=df_counts_admired, x='Country', y='Language', size='AdmirationFreq', hue='Language', sizes=(50, 2000), alpha=0.6)

plt.title('Bubble Plot for Admired Technologies Across Countries')
plt.xlabel('Country')
plt.ylabel('Admired Language')
plt.xticks(rotation=30)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Languages')
plt.tight_layout()
plt.show()
```
